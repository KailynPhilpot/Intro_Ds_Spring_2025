# Breaking down the 90s Hit Formula




The "Classic 90s Hits" dataset from Kaggle features a collection of popular songs from the 1990s, capturing key musical characteristics. It includes variables such as song title, artist, tempo, duration, and various audio features like danceability and energy.


```python
import pandas as pd

# Load dataset
df = pd.read_csv('/content/1990sClassicHits.csv')
df.head()
```

### SECTION I: DATASET OVERVIEW AND STRUCTURE
---

#### 1. Shape and Structure of the Dataset


```python
# Shape of dataset
print("Dataset Shape:", df.shape)

# First few rows
print("\nFirst 5 rows:")
display(df.head())

# Last few rows
print("\nLast 5 rows:")
display(df.tail())

# General info
print("\nData Info:\n")
df.info()
```

#### 2. Variable Types


```python
# Check data types
print("Data Types:")
print(df.dtypes)
```

#### 3. Missing Values


```python
# Total missing values per column
missing_counts = df.isnull().sum()
missing_counts = missing_counts[missing_counts > 0]

print("Missing Values:")
print(missing_counts)
print("\nNo missing values detected.")
```

There are **no missing values** in the dataset, so no imputation or data cleaning is needed for nulls.


### SECTION II: DATA CLEANING & INDEXING


---



#### 4. Check for and Remove Duplicates


```python
# Checking for duplicates
duplicates = df.duplicated().sum()
print(f"Number of duplicate rows: {duplicates}")

# Removing duplicates
df = df.drop_duplicates()

print(f"New shape after removing duplicates: {df.shape}")
```

We found 0 duplicate rows in the dataset. These were removed using .drop_duplicates(), resulting in a new shape of 919 x 17.

```python
# Convert to categorical
print(df['Time_Signature'].dtype)
print(df['Time_Signature'].unique())
df['Time_signature'] = df['Time_Signature'].astype('category')
```

Converting the Time Signature values to categorical to make it a cleaner dataset.

#### 5. Filtering Using Logical Conditions

```python
# Filtering songs that are likely vocal
vocal_songs = df[df['Instrumentalness'] < 0.2]

print(f"Number of vocal songs: {vocal_songs.shape[0]}")
vocal_songs.head()
```

I filtered songs where Instrumentalness is less than 0.2 to capture tracks that are likely to have vocals.

```python
# Filtering songs with high loudness
loud_songs = df[df['Loudness'] > -5]

print(f"Number of loud songs: {loud_songs.shape[0]}")
loud_songs.head()
```

I filtered for songs with Loudness greater than -5 dB, capturing tracks that are significantly louder than average.

### SECTION III: DESCRIPTIVE STATISTICS

#### 6. Summary Statistics for Numerical Columns

```python
# Get descriptive statistics for numerical features
df.describe()
```

Interesting insight: The average Danceability score is around .623, which suggest that 90s hits were generally usually rhythmically engaging.



#### 7. Categorical Analysis

```python
# Most common artists in the dataset
df['Artist'].value_counts().head(10)
```

The most common 3 artists were Mariah Carey, Madonna, and Jannet Jackson.

```python
# Creating tempo bins
tempo_bins = [0, 90, 110, 130, 200]
tempo_labels = ['Slow (0-90 BPM)', 'Moderate (90-110 BPM)', 'Fast (110-130 BPM)', 'Very Fast (130+ BPM)']
df['Tempo_range'] = pd.cut(df['Tempo'], bins=tempo_bins, labels=tempo_labels)

# Finding Average danceability by tempo group
df.groupby('Tempo_range')['Danceability'].mean().sort_values(ascending=False)
```

Songs in the Fast (110–130 BPM) category had the highest average Danceability, suggesting that upbeat tempos drive rhythm and movement.


#### 8. Grouped Summary Statistics

Key vs. Loudness

```python
# Average loudness by key
df.groupby('Key')['Loudness'].mean().sort_values(ascending=False)
```

This tells us which musical keys tend to have louder more intense/energetic tracks.


```python
# Count songs per time signature
df['Time_signature'].value_counts().sort_index()

```

Danceability Avg. vs. Time Signature

```python
df.groupby('Time_Signature')['Danceability'].mean()
```

We converted the Time_Signature column to categorical format to better represent musical structure. Most songs are in '4/4', which reflects the standard rhythm of 90s music.
